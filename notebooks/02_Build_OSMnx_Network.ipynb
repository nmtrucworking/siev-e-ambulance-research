# 02 — Build OSMnx Network

Notebook này là **thin wrapper** cho module chuẩn `src/build_osmnx_network.py`. Logic build graph, snapping, OD cache và QA nằm trong module để có thể chạy bằng CLI/CI.

> `RUN_NETWORK=False` chỉ chạy precheck. Đặt `True` để build/load graph và sinh `data/osmnx/*`.

````python
from pathlib import Path
import sys, json

cwd = Path.cwd().resolve()
ROOT = next((p for p in [cwd, cwd.parent, *cwd.parents] if (p/"data"/"sample").exists()), None)
if ROOT is None:
    raise FileNotFoundError("Không tìm thấy project root")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.build_osmnx_network import BuildConfig, precheck, build_network_artifacts
print("ROOT =", ROOT)
````

````python
RUN_NETWORK = False
FORCE_REDOWNLOAD = False
NETWORK_TYPE = "drive"
REVIEW_SNAP_DISTANCE_M = 250.0
BUILD_INITIAL_FLEET_INCIDENT_OD = False

cfg = BuildConfig(
    root=ROOT,
    network_type=NETWORK_TYPE,
    review_snap_distance_m=REVIEW_SNAP_DISTANCE_M,
    force_redownload=FORCE_REDOWNLOAD,
    build_initial_fleet_incident_od=BUILD_INITIAL_FLEET_INCIDENT_OD,
)
cfg
````

````python
if RUN_NETWORK:
    result = build_network_artifacts(cfg)
else:
    result = precheck(cfg)
    print("RUN_NETWORK=False → precheck only; không download/build graph.")

print(json.dumps(result, indent=2, ensure_ascii=False, default=str))
````

## CLI tương đương

Precheck:

```bash
.venv/bin/python src/build_osmnx_network.py --precheck
```

Build thật:

```bash
.venv/bin/python src/build_osmnx_network.py
```

Tải lại graph bắt buộc:

```bash
.venv/bin/python src/build_osmnx_network.py --force-redownload
```
